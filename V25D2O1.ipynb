## Oppgave 1
En bil kjører opp en spiralformet vei inne i et parkeringshus. 
Posisjonen til bilen etter t sekunder er gitt ved .
$$\begin{align*}
\vec{r}(t)=\left[4\cos\left(\frac{\pi}{5}t \right),4\sin \left(\frac{\pi}{5}t+2 \right),5+\frac{1}{3}t \right], \quad t\in[0,20]
\end{align*}$$
der bakkenivå et er xy-planet i et koordinatsystem med meter som enhet langs aksene.<br>
a) Hvor høyt over bakkenivå et er bilen etter 5 sekunder? </li>




```python
import numpy as np
pi=np.pi

def r(t):
    return np.array([4*np.cos(pi/5*t),4*np.sin(pi/5*t)+2,5+1/3*t])

print(f"Bilen er {r(5)[2]} meter over bakken")
```

b) Bestem fartsvektoren $\vec{v}(t)$
og farten til bilen etter 10 sekunder.

```python
import sympy as sp

t = sp.symbols('t')

def r(t):
    return sp.Matrix([
    4*sp.cos(sp.pi*t/5),
    4*sp.sin(sp.pi*t/5) + 2,
    5 + t/3
])

v = sp.diff(r(t), t)   # hastighetsvektor

v_func = sp.lambdify(t, v, "numpy")

v_10=np.linalg.norm(v_func(10))
print(f"v(t)=[{v[0]},{v[1]},{v[2]}]")
print(f"Farten etter 10 sekunder er {v_10}")
```
