## Eksempel 16

Et tetraeder med hjørner i $A(1,-1,2), B(4,1,6), C(2,2,2)$ og $D(3,0,-2)$.<br>
Finn volumet av tetraederet.

```python
import numpy as np
A=np.array([1,-1,2])
B=np.array([4,1,6])
C=np.array([2,2,2])
D=np.array([3,0,-2])

AB=B-A
AC=C-A
AD=D-A

ABXAC=np.cross(AB,AC)
Volum=np.linalg.norm(np.dot(ABXAC,AD))
Volum=Volum/6
print(f"Volumet er {Volum}")
```

## Eksempel 35

```python
def kule(x,y,z):
    return (x-2)**2+(y+4)**2+(z-1)**2

def sjekk(x,y,z):
    if kule(x,y,z)==9:
        return f"Punktet {x,y,z} ligger på kuleflaten"
    else:
        return f"Punktet {x,y,z} ligger ikke på kuleflaten"

[x_1,y_1,z_1]=[0,-3,3]
[x_2,y_2,z_2]=[2,2,2]

print(sjekk(x_1,y_1,z_1))
print(sjekk(x_2,y_2,z_2))


```

```python
import numpy as np

S=np.array([2,-1,4])
R=3
n=np.array([1,2,2])
d=14

def Avstand_punkt_til_plan(S,n,d):
    return abs(np.dot(n,S)-d)/np.linalg.norm(n)

D=Avstand_punkt_til_plan(S,n,d)

r=np.sqrt(R**2-D**2)

C=S-(np.dot(n,S)-d)/np.dot(n,n)*n

print(D,r,C)
```
